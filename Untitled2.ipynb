```python
# ============================================================
# NATURAL LANGUAGE PROCESSING FOR LARGE-SCALE TEXT ANALYTICS
# ============================================================
#
# Tasks:
#   1. Topic Modelling
#   2. Text Classification
#   3. Sentiment Analysis
#
# Representations:
#   - TF-IDF
#   - Word2Vec
#   - GloVe
#   - BERT-family embeddings
#
# Models:
#   - Logistic Regression
#   - Linear SVM
#   - BERT-based classifier
#
# Evaluation:
#   - Accuracy
#   - Precision
#   - Recall
#   - F1
#   - Confusion matrices
#   - Topic coherence
#   - Training time
#   - Representation comparison
#
# Datasets:
#   - 20 Newsgroups
#   - IMDb Movie Reviews
#
# ============================================================


# ============================================================
# 0. INSTALL REQUIRED PACKAGES
# ============================================================

!pip -q install gensim transformers sentencepiece datasets \
    nltk scikit-learn matplotlib seaborn pandas numpy scipy joblib


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import re
import time
import random
import warnings
import joblib

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from collections import Counter

warnings.filterwarnings("ignore")


# Scikit-learn

from sklearn.datasets import fetch_20newsgroups

from sklearn.model_selection import train_test_split

from sklearn.feature_extraction.text import (
    TfidfVectorizer,
    CountVectorizer
)

from sklearn.linear_model import LogisticRegression

from sklearn.svm import LinearSVC

from sklearn.decomposition import LatentDirichletAllocation

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)


# Gensim

from gensim.models import Word2Vec

from gensim.models import KeyedVectors


# Hugging Face

from datasets import load_dataset

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    Trainer,
    TrainingArguments
)


# PyTorch

import torch


# ============================================================
# 2. REPRODUCIBILITY
# ============================================================

SEED = 42

random.seed(SEED)

np.random.seed(SEED)

torch.manual_seed(SEED)

if torch.cuda.is_available():

    torch.cuda.manual_seed_all(SEED)


print(
    "Random seed:",
    SEED
)

print(
    "CUDA available:",
    torch.cuda.is_available()
)


# ============================================================
# 3. DIRECTORY STRUCTURE
# ============================================================

directories = [

    "data",

    "models",

    "results",

    "plots",

    "embeddings"

]

for directory in directories:

    os.makedirs(
        directory,
        exist_ok=True
    )


print(
    "\nProject directories created."
)


# ============================================================
# 4. TEXT PREPROCESSING FUNCTION
# ============================================================

def clean_text(text):

    text = str(text)

    # Lowercase

    text = text.lower()

    # Remove URLs

    text = re.sub(
        r"http\S+|www\S+",
        " ",
        text
    )

    # Remove email addresses

    text = re.sub(
        r"\S+@\S+",
        " ",
        text
    )

    # Remove HTML

    text = re.sub(
        r"<.*?>",
        " ",
        text
    )

    # Keep alphabetic characters

    text = re.sub(
        r"[^a-zA-Z\s]",
        " ",
        text
    )

    # Remove extra whitespace

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


# ============================================================
# 5. LOAD 20 NEWSGROUPS DATASET
# ============================================================

print(
    "\nLoading 20 Newsgroups dataset..."
)


newsgroups_train = fetch_20newsgroups(

    subset="train",

    remove=(
        "headers",
        "footers",
        "quotes"
    )

)


newsgroups_test = fetch_20newsgroups(

    subset="test",

    remove=(
        "headers",
        "footers",
        "quotes"
    )

)


train_texts_raw = (
    newsgroups_train.data
)

test_texts_raw = (
    newsgroups_test.data
)

y_train_news = (
    newsgroups_train.target
)

y_test_news = (
    newsgroups_test.target
)


target_names = (
    newsgroups_train.target_names
)


print(
    "Training documents:",
    len(train_texts_raw)
)

print(
    "Testing documents:",
    len(test_texts_raw)
)

print(
    "Number of classes:",
    len(target_names)
)


# ============================================================
# 6. CLEAN NEWSGROUP TEXT
# ============================================================

print(
    "\nCleaning text..."
)


X_train_news_text = [

    clean_text(text)

    for text in train_texts_raw

]


X_test_news_text = [

    clean_text(text)

    for text in test_texts_raw

]


# Remove empty documents

train_nonempty = [

    i

    for i, text in enumerate(
        X_train_news_text
    )

    if len(text) > 10

]


test_nonempty = [

    i

    for i, text in enumerate(
        X_test_news_text
    )

    if len(text) > 10

]


X_train_news_text = [

    X_train_news_text[i]

    for i in train_nonempty

]


y_train_news = (

    y_train_news[train_nonempty]

)


X_test_news_text = [

    X_test_news_text[i]

    for i in test_nonempty

]


y_test_news = (

    y_test_news[test_nonempty]

)


print(
    "Non-empty training documents:",
    len(X_train_news_text)
)

print(
    "Non-empty test documents:",
    len(X_test_news_text)
)


# ============================================================
# 7. SAVE DATASET SUMMARY
# ============================================================

dataset_summary = pd.DataFrame({

    "Dataset": [

        "20 Newsgroups",
        "20 Newsgroups"

    ],

    "Split": [

        "Train",
        "Test"

    ],

    "Documents": [

        len(X_train_news_text),
        len(X_test_news_text)

    ],

    "Classes": [

        len(target_names),
        len(target_names)

    ]

})


dataset_summary.to_csv(

    "results/dataset_summary.csv",

    index=False

)


# ============================================================
# 8. CLASS DISTRIBUTION
# ============================================================

train_class_counts = Counter(
    y_train_news
)


class_distribution = pd.DataFrame({

    "Class": [

        target_names[i]

        for i in train_class_counts.keys()

    ],

    "Count": [

        train_class_counts[i]

        for i in train_class_counts.keys()

    ]

})


class_distribution = (
    class_distribution
    .sort_values(
        "Count",
        ascending=False
    )
)


class_distribution.to_csv(

    "results/newsgroups_class_distribution.csv",

    index=False

)


plt.figure(
    figsize=(12, 7)
)


plt.barh(

    class_distribution["Class"],

    class_distribution["Count"]

)


plt.title(
    "20 Newsgroups Class Distribution"
)

plt.xlabel(
    "Number of Documents"
)

plt.ylabel(
    "Newsgroup"
)

plt.tight_layout()


plt.savefig(

    "plots/newsgroups_class_distribution.png",

    dpi=300

)


plt.show()

plt.close()


# ============================================================
# 9. TF-IDF REPRESENTATION
# ============================================================

print(
    "\nCreating TF-IDF representation..."
)


tfidf_vectorizer = TfidfVectorizer(

    max_features=30000,

    min_df=3,

    max_df=0.95,

    sublinear_tf=True,

    ngram_range=(1, 2),

    stop_words="english"

)


start_time = time.time()


X_train_tfidf = (
    tfidf_vectorizer.fit_transform(
        X_train_news_text
    )
)


X_test_tfidf = (
    tfidf_vectorizer.transform(
        X_test_news_text
    )
)


tfidf_time = (
    time.time() - start_time
)


print(
    "TF-IDF training matrix:",
    X_train_tfidf.shape
)

print(
    "TF-IDF creation time:",
    round(
        tfidf_time,
        2
    ),
    "seconds"
)


# Save vectorizer

joblib.dump(

    tfidf_vectorizer,

    "models/tfidf_vectorizer.joblib"

)


# ============================================================
# 10. TF-IDF CLASSIFICATION - LOGISTIC REGRESSION
# ============================================================

print(
    "\nTraining TF-IDF Logistic Regression..."
)


start_time = time.time()


tfidf_lr = LogisticRegression(

    max_iter=1000,

    C=2,

    random_state=SEED

)


tfidf_lr.fit(

    X_train_tfidf,

    y_train_news

)


tfidf_lr_training_time = (
    time.time() - start_time
)


tfidf_lr_predictions = (
    tfidf_lr.predict(
        X_test_tfidf
    )
)


# Save model

joblib.dump(

    tfidf_lr,

    "models/tfidf_logistic_regression.joblib"

)


# ============================================================
# 11. TF-IDF METRICS
# ============================================================

tfidf_lr_accuracy = accuracy_score(

    y_test_news,

    tfidf_lr_predictions

)


tfidf_lr_precision = precision_score(

    y_test_news,

    tfidf_lr_predictions,

    average="weighted",

    zero_division=0

)


tfidf_lr_recall = recall_score(

    y_test_news,

    tfidf_lr_predictions,

    average="weighted",

    zero_division=0

)


tfidf_lr_f1 = f1_score(

    y_test_news,

    tfidf_lr_predictions,

    average="weighted",

    zero_division=0

)


print(
    "\nTF-IDF Logistic Regression:"
)

print(
    "Accuracy:",
    round(
        tfidf_lr_accuracy,
        4
    )
)

print(
    "F1:",
    round(
        tfidf_lr_f1,
        4
    )
)


# ============================================================
# 12. TF-IDF CLASSIFICATION - LINEAR SVM
# ============================================================

print(
    "\nTraining TF-IDF Linear SVM..."
)


start_time = time.time()


tfidf_svm = LinearSVC(

    C=1.5,

    random_state=SEED

)


tfidf_svm.fit(

    X_train_tfidf,

    y_train_news

)


tfidf_svm_training_time = (
    time.time() - start_time
)


tfidf_svm_predictions = (
    tfidf_svm.predict(
        X_test_tfidf
    )
)


joblib.dump(

    tfidf_svm,

    "models/tfidf_linear_svm.joblib"

)


tfidf_svm_accuracy = accuracy_score(

    y_test_news,

    tfidf_svm_predictions

)


tfidf_svm_precision = precision_score(

    y_test_news,

    tfidf_svm_predictions,

    average="weighted",

    zero_division=0

)


tfidf_svm_recall = recall_score(

    y_test_news,

    tfidf_svm_predictions,

    average="weighted",

    zero_division=0

)


tfidf_svm_f1 = f1_score(

    y_test_news,

    tfidf_svm_predictions,

    average="weighted",

    zero_division=0

)


print(
    "TF-IDF SVM Accuracy:",
    round(
        tfidf_svm_accuracy,
        4
    )
)

print(
    "TF-IDF SVM F1:",
    round(
        tfidf_svm_f1,
        4
    )
)


# ============================================================
# 13. WORD2VEC TRAINING CORPUS
# ============================================================

print(
    "\nPreparing Word2Vec corpus..."
)


tokenized_train = [

    text.split()

    for text in X_train_news_text

]


# ============================================================
# 14. TRAIN WORD2VEC
# ============================================================

print(
    "Training Word2Vec..."
)


start_time = time.time()


word2vec_model = Word2Vec(

    sentences=tokenized_train,

    vector_size=200,

    window=5,

    min_count=3,

    workers=4,

    sg=1,

    epochs=10,

    seed=SEED

)


word2vec_training_time = (
    time.time() - start_time
)


word2vec_model.save(

    "models/word2vec.model"

)


print(
    "Word2Vec vocabulary:",
    len(
        word2vec_model.wv
    )
)

print(
    "Word2Vec training time:",
    round(
        word2vec_training_time,
        2
    ),
    "seconds"
)


# ============================================================
# 15. WORD2VEC DOCUMENT EMBEDDINGS
# ============================================================

def document_word2vec_embedding(

    text,

    model

):

    words = text.split()

    vectors = []

    for word in words:

        if word in model.wv:

            vectors.append(
                model.wv[word]
            )

    if len(vectors) == 0:

        return np.zeros(
            model.vector_size
        )

    return np.mean(
        vectors,
        axis=0
    )


print(
    "\nCreating Word2Vec document embeddings..."
)


X_train_w2v = np.vstack([

    document_word2vec_embedding(

        text,

        word2vec_model

    )

    for text in X_train_news_text

])


X_test_w2v = np.vstack([

    document_word2vec_embedding(

        text,

        word2vec_model

    )

    for text in X_test_news_text

])


print(
    "Word2Vec training embeddings:",
    X_train_w2v.shape
)


# ============================================================
# 16. WORD2VEC CLASSIFICATION
# ============================================================

print(
    "\nTraining Word2Vec classifier..."
)


start_time = time.time()


w2v_classifier = LogisticRegression(

    max_iter=1000,

    C=2,

    random_state=SEED

)


w2v_classifier.fit(

    X_train_w2v,

    y_train_news

)


w2v_training_time = (
    time.time() - start_time
)


w2v_predictions = (
    w2v_classifier.predict(
        X_test_w2v
    )
)


joblib.dump(

    w2v_classifier,

    "models/word2vec_logistic_regression.joblib"

)


w2v_accuracy = accuracy_score(

    y_test_news,

    w2v_predictions

)


w2v_precision = precision_score(

    y_test_news,

    w2v_predictions,

    average="weighted",

    zero_division=0

)


w2v_recall = recall_score(

    y_test_news,

    w2v_predictions,

    average="weighted",

    zero_division=0

)


w2v_f1 = f1_score(

    y_test_news,

    w2v_predictions,

    average="weighted",

    zero_division=0

)


print(
    "Word2Vec Accuracy:",
    round(
        w2v_accuracy,
        4
    )
)

print(
    "Word2Vec F1:",
    round(
        w2v_f1,
        4
    )
)


# ============================================================
# 17. GLOVE EMBEDDINGS
# ============================================================
#
# Download GloVe 6B 100-dimensional embeddings.
#
# ============================================================

print(
    "\nDownloading GloVe embeddings..."
)


!wget -q --show-progress \
    https://nlp.stanford.edu/data/glove.6B.zip \
    -O glove.6B.zip


!unzip -q -o glove.6B.zip


# ============================================================
# 18. LOAD GLOVE
# ============================================================

glove_file = "glove.6B.100d.txt"


print(
    "\nLoading GloVe..."
)


glove_vectors = {}


with open(

    glove_file,

    encoding="utf-8"

) as file:

    for line in file:

        parts = line.split()

        word = parts[0]

        vector = np.asarray(

            parts[1:],

            dtype="float32"

        )

        glove_vectors[word] = vector


print(
    "GloVe vocabulary loaded:",
    len(glove_vectors)
)


# ============================================================
# 19. GLOVE DOCUMENT EMBEDDINGS
# ============================================================

def document_glove_embedding(

    text,

    glove

):

    words = text.split()

    vectors = []

    for word in words:

        if word in glove:

            vectors.append(
                glove[word]
            )

    if len(vectors) == 0:

        return np.zeros(100)

    return np.mean(
        vectors,
        axis=0
    )


print(
    "\nCreating GloVe document embeddings..."
)


X_train_glove = np.vstack([

    document_glove_embedding(

        text,

        glove_vectors

    )

    for text in X_train_news_text

])


X_test_glove = np.vstack([

    document_glove_embedding(

        text,

        glove_vectors

    )

    for text in X_test_news_text

])


print(
    "GloVe training embeddings:",
    X_train_glove.shape
)


# ============================================================
# 20. GLOVE CLASSIFICATION
# ============================================================

print(
    "\nTraining GloVe classifier..."
)


start_time = time.time()


glove_classifier = LogisticRegression(

    max_iter=1000,

    C=2,

    random_state=SEED

)


glove_classifier.fit(

    X_train_glove,

    y_train_news

)


glove_training_time = (
    time.time() - start_time
)


glove_predictions = (
    glove_classifier.predict(
        X_test_glove
    )
)


joblib.dump(

    glove_classifier,

    "models/glove_logistic_regression.joblib"

)


glove_accuracy = accuracy_score(

    y_test_news,

    glove_predictions

)


glove_precision = precision_score(

    y_test_news,

    glove_predictions,

    average="weighted",

    zero_division=0

)


glove_recall = recall_score(

    y_test_news,

    glove_predictions,

    average="weighted",

    zero_division=0

)


glove_f1 = f1_score(

    y_test_news,

    glove_predictions,

    average="weighted",

    zero_division=0

)


print(
    "GloVe Accuracy:",
    round(
        glove_accuracy,
        4
    )
)

print(
    "GloVe F1:",
    round(
        glove_f1,
        4
    )
)


# ============================================================
# 21. TOPIC MODELLING
# ============================================================

print(
    "\nStarting topic modelling..."
)


# Use CountVectorizer for LDA

count_vectorizer = CountVectorizer(

    max_features=15000,

    min_df=5,

    max_df=0.95,

    stop_words="english"

)


X_count = count_vectorizer.fit_transform(

    X_train_news_text

)


print(
    "LDA input matrix:",
    X_count.shape
)


# ============================================================
# 22. LDA MODEL
# ============================================================

N_TOPICS = 20


lda_model = LatentDirichletAllocation(

    n_components=N_TOPICS,

    random_state=SEED,

    learning_method="online",

    batch_size=128,

    max_iter=20,

    n_jobs=-1

)


start_time = time.time()


lda_model.fit(
    X_count
)


lda_training_time = (
    time.time() - start_time
)


joblib.dump(

    lda_model,

    "models/lda_topic_model.joblib"

)


joblib.dump(

    count_vectorizer,

    "models/count_vectorizer.joblib"

)


print(
    "LDA training time:",
    round(
        lda_training_time,
        2
    ),
    "seconds"
)


# ============================================================
# 23. DISPLAY TOP WORDS PER TOPIC
# ============================================================

feature_names = (
    count_vectorizer
    .get_feature_names_out()
)


topic_records = []


for topic_idx, topic in enumerate(

    lda_model.components_

):

    top_indices = topic.argsort()[
        -15:
    ][::-1]


    top_words = [

        feature_names[i]

        for i in top_indices

    ]


    topic_records.append({

        "Topic":
            topic_idx + 1,

        "Top_Words":
            ", ".join(top_words)

    })


topic_df = pd.DataFrame(
    topic_records
)


print(
    "\nTop words per topic:"
)

print(
    topic_df.to_string(
        index=False
    )
)


topic_df.to_csv(

    "results/lda_topics.csv",

    index=False

)


# ============================================================
# 24. TOPIC WORD PLOT
# ============================================================

for topic_idx, topic in enumerate(

    lda_model.components_

):

    top_indices = topic.argsort()[
        -10:
    ][::-1]


    top_words = [

        feature_names[i]

        for i in top_indices

    ]


    top_weights = [

        topic[i]

        for i in top_indices

    ]


    plt.figure(
        figsize=(9, 5)
    )


    plt.barh(

        top_words[::-1],

        top_weights[::-1]

    )


    plt.title(

        f"Top Words - Topic {topic_idx + 1}"

    )

    plt.xlabel(
        "Topic Weight"
    )

    plt.ylabel(
        "Word"
    )


    plt.tight_layout()


    plt.savefig(

        f"plots/topic_{topic_idx + 1}_words.png",

        dpi=250

    )


    plt.close()


# ============================================================
# 25. TOPIC DISTRIBUTION
# ============================================================

document_topics = lda_model.transform(
    X_count
)


dominant_topics = (
    np.argmax(
        document_topics,
        axis=1
    )
)


topic_distribution = (
    pd.Series(
        dominant_topics
    )
    .value_counts()
    .sort_index()
)


plt.figure(
    figsize=(12, 6)
)


plt.bar(

    topic_distribution.index + 1,

    topic_distribution.values

)


plt.title(
    "Dominant Topic Distribution"
)

plt.xlabel(
    "Topic"
)

plt.ylabel(
    "Number of Documents"
)


plt.tight_layout()


plt.savefig(

    "plots/topic_distribution.png",

    dpi=300

)


plt.show()

plt.close()


# ============================================================
# 26. SIMPLE TOPIC COHERENCE SCORE
# ============================================================
#
# This calculates a simple word co-occurrence coherence
# approximation.
#
# Higher values indicate stronger word association.
#
# ============================================================

def topic_coherence_score(

    topic_words,

    documents

):

    document_sets = [

        set(
            doc.split()
        )

        for doc in documents

    ]


    scores = []


    for i in range(

        len(topic_words)

    ):

        for j in range(

            i + 1,

            len(topic_words)

        ):

            word_i = topic_words[i]

            word_j = topic_words[j]


            count_i = sum(

                word_i in doc

                for doc in document_sets

            )


            if count_i == 0:

                continue


            cooccur = sum(

                (

                    word_i in doc

                    and

                    word_j in doc

                )

                for doc in document_sets

            )


            score = (

                cooccur / count_i

            )


            scores.append(
                score
            )


    if len(scores) == 0:

        return 0

    return np.mean(
        scores
    )


coherence_records = []


# Use a subset for speed

coherence_documents = X_train_news_text[
    :5000
]


for topic_idx, topic in enumerate(

    lda_model.components_

):

    top_indices = topic.argsort()[
        -10:
    ][::-1]


    top_words = [

        feature_names[i]

        for i in top_indices

    ]


    coherence = (
        topic_coherence_score(
            top_words,
            coherence_documents
        )
    )


    coherence_records.append({

        "Topic":
            topic_idx + 1,

        "Coherence":
            coherence

    })


coherence_df = pd.DataFrame(
    coherence_records
)


coherence_df.to_csv(

    "results/topic_coherence.csv",

    index=False

)


print(
    "\nAverage topic coherence:",
    round(
        coherence_df["Coherence"].mean(),
        4
    )
)


# ============================================================
# 27. LOAD IMDB SENTIMENT DATASET
# ============================================================

print(
    "\nLoading IMDb sentiment dataset..."
)


imdb = load_dataset(
    "stanfordnlp/imdb"
)


print(
    imdb
)


# ============================================================
# 28. CONVERT IMDB TO PANDAS
# ============================================================

imdb_train = pd.DataFrame(

    imdb["train"]

)


imdb_test = pd.DataFrame(

    imdb["test"]

)


print(
    "IMDb training documents:",
    len(imdb_train)
)

print(
    "IMDb testing documents:",
    len(imdb_test)
)


# ============================================================
# 29. CLEAN IMDb DATA
# ============================================================

imdb_train["text_clean"] = (

    imdb_train["text"]

    .apply(
        clean_text
    )

)


imdb_test["text_clean"] = (

    imdb_test["text"]

    .apply(
        clean_text
    )

)


# ============================================================
# 30. LIMIT DATA FOR CPU-FRIENDLY RUNNING
# ============================================================
#
# Set to None if you want to use the complete dataset.
#
# ============================================================

MAX_IMDB_TRAIN = 20000

MAX_IMDB_TEST = 5000


imdb_train = imdb_train.sample(

    n=min(
        MAX_IMDB_TRAIN,
        len(imdb_train)
    ),

    random_state=SEED

)


imdb_test = imdb_test.sample(

    n=min(
        MAX_IMDB_TEST,
        len(imdb_test)
    ),

    random_state=SEED

)


X_imdb_train = (
    imdb_train["text_clean"]
    .tolist()
)

y_imdb_train = (
    imdb_train["label"]
    .values
)


X_imdb_test = (
    imdb_test["text_clean"]
    .tolist()
)

y_imdb_test = (
    imdb_test["label"]
    .values
)


# ============================================================
# 31. IMDb TF-IDF
# ============================================================

print(
    "\nCreating IMDb TF-IDF features..."
)


imdb_tfidf = TfidfVectorizer(

    max_features=30000,

    min_df=3,

    max_df=0.95,

    ngram_range=(1, 2),

    sublinear_tf=True,

    stop_words="english"

)


X_imdb_tfidf_train = (

    imdb_tfidf.fit_transform(
        X_imdb_train
    )

)


X_imdb_tfidf_test = (

    imdb_tfidf.transform(
        X_imdb_test
    )

)


joblib.dump(

    imdb_tfidf,

    "models/imdb_tfidf_vectorizer.joblib"

)


# ============================================================
# 32. IMDb SENTIMENT CLASSIFIER
# ============================================================

print(
    "\nTraining IMDb sentiment classifier..."
)


start_time = time.time()


imdb_classifier = LogisticRegression(

    max_iter=1000,

    C=2,

    random_state=SEED

)


imdb_classifier.fit(

    X_imdb_tfidf_train,

    y_imdb_train

)


imdb_training_time = (
    time.time() - start_time
)


imdb_predictions = (

    imdb_classifier.predict(
        X_imdb_tfidf_test
    )

)


joblib.dump(

    imdb_classifier,

    "models/imdb_sentiment_classifier.joblib"

)


imdb_accuracy = accuracy_score(

    y_imdb_test,

    imdb_predictions

)


imdb_precision = precision_score(

    y_imdb_test,

    imdb_predictions,

    zero_division=0

)


imdb_recall = recall_score(

    y_imdb_test,

    imdb_predictions,

    zero_division=0

)


imdb_f1 = f1_score(

    y_imdb_test,

    imdb_predictions,

    zero_division=0

)


print(
    "\nIMDb sentiment results:"
)

print(
    "Accuracy:",
    round(
        imdb_accuracy,
        4
    )
)

print(
    "F1:",
    round(
        imdb_f1,
        4
    )
)


# ============================================================
# 33. IMDb CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    y_imdb_test,

    imdb_predictions

)


plt.figure(
    figsize=(6, 5)
)


sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=[
        "Negative",
        "Positive"
    ],

    yticklabels=[
        "Negative",
        "Positive"
    ]

)


plt.title(
    "IMDb Sentiment Confusion Matrix"
)

plt.xlabel(
    "Predicted"
)

plt.ylabel(
    "Actual"
)


plt.tight_layout()


plt.savefig(

    "plots/imdb_confusion_matrix.png",

    dpi=300

)


plt.show()

plt.close()


# ============================================================
# 34. BERT MODEL
# ============================================================
#
# BERT is computationally expensive.
#
# The code automatically uses GPU when available.
#
# ============================================================

print(
    "\nPreparing BERT sentiment model..."
)


MODEL_NAME = (
    "distilbert-base-uncased"
)


tokenizer = AutoTokenizer.from_pretrained(

    MODEL_NAME

)


bert_model = AutoModelForSequenceClassification.from_pretrained(

    MODEL_NAME,

    num_labels=2

)


# ============================================================
# 35. CREATE SMALL BERT DATASET
# ============================================================
#
# Increase these values if you have sufficient GPU resources.
#
# ============================================================

BERT_TRAIN_SIZE = 5000

BERT_TEST_SIZE = 1000


bert_train_df = imdb_train.sample(

    n=min(
        BERT_TRAIN_SIZE,
        len(imdb_train)
    ),

    random_state=SEED

)


bert_test_df = imdb_test.sample(

    n=min(
        BERT_TEST_SIZE,
        len(imdb_test)
    ),

    random_state=SEED

)


# ============================================================
# 36. TOKENIZATION
# ============================================================

def tokenize_texts(

    texts,

    labels

):

    encoded = tokenizer(

        texts,

        truncation=True,

        padding=True,

        max_length=256

    )


    encoded["labels"] = list(
        labels
    )


    return encoded


bert_train_encoded = tokenize_texts(

    bert_train_df["text"].tolist(),

    bert_train_df["label"].tolist()

)


bert_test_encoded = tokenize_texts(

    bert_test_df["text"].tolist(),

    bert_test_df["label"].tolist()

)


# ============================================================
# 37. HUGGING FACE DATASET
# ============================================================

from datasets import Dataset


bert_train_dataset = Dataset.from_dict(

    bert_train_encoded

)


bert_test_dataset = Dataset.from_dict(

    bert_test_encoded

)


# ============================================================
# 38. BERT TRAINING CONFIGURATION
# ============================================================

training_args = TrainingArguments(

    output_dir="models/bert_output",

    eval_strategy="epoch",

    save_strategy="epoch",

    learning_rate=2e-5,

    per_device_train_batch_size=8,

    per_device_eval_batch_size=8,

    num_train_epochs=2,

    weight_decay=0.01,

    logging_dir="results/bert_logs",

    logging_steps=100,

    load_best_model_at_end=True,

    metric_for_best_model="f1",

    report_to="none",

    seed=SEED

)


# ============================================================
# 39. BERT METRICS
# ============================================================

def compute_bert_metrics(

    eval_prediction

):

    logits, labels = eval_prediction

    predictions = np.argmax(

        logits,

        axis=-1

    )


    return {

        "accuracy":
            accuracy_score(
                labels,
                predictions
            ),

        "precision":
            precision_score(
                labels,
                predictions,
                zero_division=0
            ),

        "recall":
            recall_score(
                labels,
                predictions,
                zero_division=0
            ),

        "f1":
            f1_score(
                labels,
                predictions,
                zero_division=0
            )

    }


# ============================================================
# 40. BERT TRAINER
# ============================================================

trainer = Trainer(

    model=bert_model,

    args=training_args,

    train_dataset=bert_train_dataset,

    eval_dataset=bert_test_dataset,

    tokenizer=tokenizer,

    compute_metrics=compute_bert_metrics

)


# ============================================================
# 41. TRAIN BERT
# ============================================================

print(
    "\nTraining DistilBERT..."
)

print(
    "This may take several minutes."
)


start_time = time.time()


trainer.train()


bert_training_time = (
    time.time() - start_time
)


# ============================================================
# 42. EVALUATE BERT
# ============================================================

bert_results = trainer.evaluate()


print(
    "\nBERT results:"
)

print(
    bert_results
)


# ============================================================
# 43. SAVE BERT MODEL
# ============================================================

trainer.save_model(

    "models/distilbert_sentiment"

)


tokenizer.save_pretrained(

    "models/distilbert_sentiment"

)


# ============================================================
# 44. BERT PREDICTIONS
# ============================================================

bert_prediction_output = (

    trainer.predict(
        bert_test_dataset
    )

)


bert_logits = (
    bert_prediction_output.predictions
)


bert_predictions = np.argmax(

    bert_logits,

    axis=1

)


bert_accuracy = accuracy_score(

    bert_test_df["label"],

    bert_predictions

)


bert_precision = precision_score(

    bert_test_df["label"],

    bert_predictions,

    zero_division=0

)


bert_recall = recall_score(

    bert_test_df["label"],

    bert_predictions,

    zero_division=0

)


bert_f1 = f1_score(

    bert_test_df["label"],

    bert_predictions,

    zero_division=0

)


print(
    "\nFinal BERT metrics:"
)

print(
    "Accuracy:",
    round(
        bert_accuracy,
        4
    )
)

print(
    "Precision:",
    round(
        bert_precision,
        4
    )
)

print(
    "Recall:",
    round(
        bert_recall,
        4
    )
)

print(
    "F1:",
    round(
        bert_f1,
        4
    )
)


# ============================================================
# 45. BERT CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    bert_test_df["label"],

    bert_predictions

)


plt.figure(
    figsize=(6, 5)
)


sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=[
        "Negative",
        "Positive"
    ],

    yticklabels=[
        "Negative",
        "Positive"
    ]

)


plt.title(
    "DistilBERT Sentiment Confusion Matrix"
)

plt.xlabel(
    "Predicted"
)

plt.ylabel(
    "Actual"
)


plt.tight_layout()


plt.savefig(

    "plots/bert_confusion_matrix.png",

    dpi=300

)


plt.show()

plt.close()


# ============================================================
# 46. TEXT REPRESENTATION COMPARISON
# ============================================================

representation_results = pd.DataFrame({

    "Representation": [

        "TF-IDF",

        "Word2Vec",

        "GloVe",

        "DistilBERT"

    ],

    "Task": [

        "20 Newsgroups Classification",

        "20 Newsgroups Classification",

        "20 Newsgroups Classification",

        "IMDb Sentiment"

    ],

    "Accuracy": [

        tfidf_lr_accuracy,

        w2v_accuracy,

        glove_accuracy,

        bert_accuracy

    ],

    "Precision": [

        tfidf_lr_precision,

        w2v_precision,

        glove_precision,

        bert_precision

    ],

    "Recall": [

        tfidf_lr_recall,

        w2v_recall,

        glove_recall,

        bert_recall

    ],

    "F1": [

        tfidf_lr_f1,

        w2v_f1,

        glove_f1,

        bert_f1

    ],

    "Training_Time_Seconds": [

        tfidf_lr_training_time,

        w2v_training_time,

        glove_training_time,

        bert_training_time

    ]

})


representation_results.to_csv(

    "results/representation_comparison.csv",

    index=False

)


print(
    "\nRepresentation comparison:"
)

print(
    representation_results
)


# ============================================================
# 47. TF-IDF VS WORD2VEC VS GLOVE PLOT
# ============================================================

traditional_comparison = pd.DataFrame({

    "Representation": [

        "TF-IDF",

        "Word2Vec",

        "GloVe"

    ],

    "Accuracy": [

        tfidf_lr_accuracy,

        w2v_accuracy,

        glove_accuracy

    ],

    "F1": [

        tfidf_lr_f1,

        w2v_f1,

        glove_f1

    ]

})


traditional_comparison = (
    traditional_comparison
    .set_index(
        "Representation"
    )
)


traditional_comparison.plot(

    kind="bar",

    figsize=(10, 6)

)


plt.title(
    "Distributional Representation Comparison"
)

plt.ylabel(
    "Score"
)

plt.xlabel(
    "Representation"
)

plt.ylim(
    0,
    1
)

plt.xticks(
    rotation=0
)

plt.tight_layout()


plt.savefig(

    "plots/traditional_representation_comparison.png",

    dpi=300

)


plt.show()

plt.close()


# ============================================================
# 48. CONTEXTUAL VS DISTRIBUTIONAL REPRESENTATIONS
# ============================================================

contextual_comparison = pd.DataFrame({

    "Representation": [

        "TF-IDF",

        "Word2Vec",

        "GloVe",

        "DistilBERT"

    ],

    "F1": [

        tfidf_lr_f1,

        w2v_f1,

        glove_f1,

        bert_f1

    ]

})


plt.figure(
    figsize=(10, 6)
)


plt.bar(

    contextual_comparison[
        "Representation"
    ],

    contextual_comparison[
        "F1"
    ]

)


plt.title(

    "Contextual vs Distributional Representation Performance"

)

plt.xlabel(
    "Representation"
)

plt.ylabel(
    "F1 Score"
)

plt.ylim(
    0,
    1
)


plt.tight_layout()


plt.savefig(

    "plots/contextual_vs_distributional.png",

    dpi=300

)


plt.show()

plt.close()


# ============================================================
# 49. TRAINING TIME COMPARISON
# ============================================================

plt.figure(
    figsize=(10, 6)
)


plt.bar(

    representation_results[
        "Representation"
    ],

    representation_results[
        "Training_Time_Seconds"
    ]

)


plt.title(
    "Representation Training Time Comparison"
)

plt.xlabel(
    "Representation"
)

plt.ylabel(
    "Training Time (seconds)"
)


plt.tight_layout()


plt.savefig(

    "plots/training_time_comparison.png",

    dpi=300

)


plt.show()

plt.close()


# ============================================================
# 50. CONFUSION MATRIX - TF-IDF NEWSGROUPS
# ============================================================

cm = confusion_matrix(

    y_test_news,

    tfidf_lr_predictions

)


plt.figure(
    figsize=(12, 10)
)


sns.heatmap(

    cm,

    cmap="Blues",

    xticklabels=target_names,

    yticklabels=target_names

)


plt.title(
    "TF-IDF Logistic Regression - 20 Newsgroups"
)

plt.xlabel(
    "Predicted Class"
)

plt.ylabel(
    "Actual Class"
)

plt.xticks(
    rotation=90
)

plt.yticks(
    rotation=0
)

plt.tight_layout()


plt.savefig(

    "plots/tfidf_newsgroups_confusion_matrix.png",

    dpi=300

)


plt.show()

plt.close()


# ============================================================
# 51. CLASSIFICATION REPORTS
# ============================================================

classification_reports = {}


classification_reports[
    "TF-IDF"
] = classification_report(

    y_test_news,

    tfidf_lr_predictions,

    target_names=target_names,

    zero_division=0

)


classification_reports[
    "Word2Vec"
] = classification_report(

    y_test_news,

    w2v_predictions,

    target_names=target_names,

    zero_division=0

)


classification_reports[
    "GloVe"
] = classification_report(

    y_test_news,

    glove_predictions,

    target_names=target_names,

    zero_division=0

)


classification_reports[
    "IMDb TF-IDF"
] = classification_report(

    y_imdb_test,

    imdb_predictions,

    zero_division=0

)


classification_reports[
    "DistilBERT"
] = classification_report(

    bert_test_df["label"],

    bert_predictions,

    zero_division=0

)


with open(

    "results/classification_reports.txt",

    "w"

) as file:

    for name, report in (
        classification_reports.items()
    ):

        file.write(

            "\n"

            + "=" * 70

            + "\n"

        )

        file.write(
            name
        )

        file.write(
            "\n"
        )

        file.write(

            "=" * 70

            + "\n"

        )

        file.write(
            report
        )

        file.write(
            "\n"
        )


# ============================================================
# 52. SAVE TEST PREDICTIONS
# ============================================================

news_predictions = pd.DataFrame({

    "Text": X_test_news_text,

    "Actual_Class":
        y_test_news,

    "TFIDF_Prediction":
        tfidf_lr_predictions,

    "Word2Vec_Prediction":
        w2v_predictions,

    "GloVe_Prediction":
        glove_predictions

})


news_predictions.to_csv(

    "results/newsgroups_predictions.csv",

    index=False

)


imdb_predictions_df = pd.DataFrame({

    "Text":
        X_imdb_test,

    "Actual":
        y_imdb_test,

    "TFIDF_Prediction":
        imdb_predictions

})


imdb_predictions_df.to_csv(

    "results/imdb_predictions.csv",

    index=False

)


# ============================================================
# 53. FINAL PROJECT SUMMARY
# ============================================================

summary = pd.DataFrame({

    "Experiment": [

        "20 Newsgroups - TF-IDF",

        "20 Newsgroups - Word2Vec",

        "20 Newsgroups - GloVe",

        "IMDb - TF-IDF",

        "IMDb - DistilBERT"

    ],

    "Representation": [

        "TF-IDF",

        "Word2Vec",

        "GloVe",

        "TF-IDF",

        "DistilBERT"

    ],

    "Task": [

        "Topic Classification",

        "Topic Classification",

        "Topic Classification",

        "Sentiment Classification",

        "Sentiment Classification"

    ],

    "Accuracy": [

        tfidf_lr_accuracy,

        w2v_accuracy,

        glove_accuracy,

        imdb_accuracy,

        bert_accuracy

    ],

    "Precision": [

        tfidf_lr_precision,

        w2v_precision,

        glove_precision,

        imdb_precision,

        bert_precision

    ],

    "Recall": [

        tfidf_lr_recall,

        w2v_recall,

        glove_recall,

        imdb_recall,

        bert_recall

    ],

    "F1": [

        tfidf_lr_f1,

        w2v_f1,

        glove_f1,

        imdb_f1,

        bert_f1

    ],

    "Training_Time_Seconds": [

        tfidf_lr_training_time,

        w2v_training_time,

        glove_training_time,

        imdb_training_time,

        bert_training_time

    ]

})


summary.to_csv(

    "results/final_project_results.csv",

    index=False

)


# ============================================================
# 54. SAVE EXPERIMENT CONFIGURATION
# ============================================================

configuration = pd.DataFrame({

    "Parameter": [

        "Random Seed",

        "TF-IDF Max Features",

        "Word2Vec Vector Size",

        "Word2Vec Window",

        "Word2Vec Epochs",

        "GloVe Dimensions",

        "LDA Topics",

        "LDA Max Iterations",

        "BERT Model",

        "BERT Max Length",

        "BERT Training Samples",

        "BERT Test Samples"

    ],

    "Value": [

        SEED,

        30000,

        200,

        5,

        10,

        100,

        N_TOPICS,

        20,

        MODEL_NAME,

        256,

        BERT_TRAIN_SIZE,

        BERT_TEST_SIZE

    ]

})


configuration.to_csv(

    "results/experiment_configuration.csv",

    index=False

)


# ============================================================
# 55. WRITE PROJECT SUMMARY
# ============================================================

with open(

    "results/project_summary.txt",

    "w"

) as file:

    file.write(

        "NATURAL LANGUAGE PROCESSING "
        "FOR LARGE-SCALE TEXT ANALYTICS\n"

    )

    file.write(
        "=" * 70
        + "\n\n"
    )

    file.write(
        "Datasets:\n"
    )

    file.write(
        "- 20 Newsgroups\n"
    )

    file.write(
        "- IMDb Movie Reviews\n\n"
    )

    file.write(
        "Representations:\n"
    )

    file.write(
        "- TF-IDF\n"
    )

    file.write(
        "- Word2Vec\n"
    )

    file.write(
        "- GloVe\n"
    )

    file.write(
        "- DistilBERT\n\n"
    )

    file.write(
        "Tasks:\n"
    )

    file.write(
        "- Topic modelling\n"
    )

    file.write(
        "- Text classification\n"
    )

    file.write(
        "- Sentiment analysis\n\n"
    )

    file.write(
        "Topic modelling:\n"
    )

    file.write(
        f"- Number of topics: {N_TOPICS}\n"
    )

    file.write(
        f"- Mean coherence: "
        f"{coherence_df['Coherence'].mean():.4f}\n\n"
    )

    file.write(
        "Classification results:\n\n"
    )

    file.write(
        summary.to_string(
            index=False
        )
    )

    file.write(
        "\n\n"
    )

    file.write(
        "Interpretation:\n"
    )

    file.write(

        "The experiment compares sparse lexical "
        "TF-IDF representations, distributional "
        "Word2Vec and GloVe embeddings, and a "
        "contextual transformer representation. "
        "The results can be used to investigate "
        "whether contextual representations provide "
        "consistent downstream performance gains "
        "relative to conventional lexical and "
        "distributional approaches.\n"

    )


# ============================================================
# 56. FINAL OUTPUT SUMMARY
# ============================================================

print(
    "\n"
    + "=" * 80
)

print(
    "NLP PROJECT COMPLETED"
)

print(
    "=" * 80
)


print(
    "\nDatasets:"
)

print(
    "- 20 Newsgroups"
)

print(
    "- IMDb Movie Reviews"
)


print(
    "\nRepresentations:"
)

print(
    "- TF-IDF"
)

print(
    "- Word2Vec"
)

print(
    "- GloVe"
)

print(
    "- DistilBERT"
)


print(
    "\nTasks:"
)

print(
    "- Topic modelling"
)

print(
    "- Text classification"
)

print(
    "- Sentiment analysis"
)


print(
    "\nGenerated directories:"
)

print(
    """
data/
models/
results/
plots/
embeddings/
"""
)


print(
    "\nImportant result files:"
)

print(
    """
results/final_project_results.csv
results/representation_comparison.csv
results/lda_topics.csv
results/topic_coherence.csv
results/classification_reports.txt
results/dataset_summary.csv
results/project_summary.txt
results/experiment_configuration.csv
results/newsgroups_predictions.csv
results/imdb_predictions.csv
"""
)


print(
    "\nSaved models:"
)

print(
    """
models/tfidf_vectorizer.joblib
models/tfidf_logistic_regression.joblib
models/tfidf_linear_svm.joblib
models/word2vec.model
models/word2vec_logistic_regression.joblib
models/glove_logistic_regression.joblib
models/lda_topic_model.joblib
models/count_vectorizer.joblib
models/imdb_tfidf_vectorizer.joblib
models/imdb_sentiment_classifier.joblib
models/distilbert_sentiment/
"""
)


print(
    "\nAll experiments completed."
)
```
